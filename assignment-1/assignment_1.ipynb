#### 8. Among whom is the proportion of those who earn a lot (>50K) greater: married or single men (marital-status feature)? Consider as married those who have a marital-status starting with Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), the rest are considered bachelors.

```python
erkekler = df[df['sex'] == 'Male']
evli_mi = erkekler['marital-status'].str.startswith('Married')
evli_orani = erkekler[evli_mi]['salary'].apply(lambda x: x == '>50K').mean()
bekar_orani = erkekler[~evli_mi]['salary'].apply(lambda x: x == '>50K').mean()
print('Evli Erkeklerin 50K Üstü Kazanma Oranı:', evli_orani)
print('Bekar Erkeklerin 50K Üstü Kazanma Oranı:', bekar_orani)
```

#### 9. What is the maximum number of hours a person works per week (hours-per-week feature)? How many people work such a number of hours, and what is the percentage of those who earn a lot (>50K) among them?

```python
max_saat = df['hours-per-week'].max()
print('Maksimum Çalışma Saati:', max_saat)
max_calisanlar = df[df['hours-per-week'] == max_saat]
kisi_sayisi = len(max_calisanlar)
print('Bu saatte çalışan kişi sayısı:', kisi_sayisi)
oran = (max_calisanlar['salary'] == '>50K').mean()
print('Yüksek Kazananların Oranı:', oran)
```
